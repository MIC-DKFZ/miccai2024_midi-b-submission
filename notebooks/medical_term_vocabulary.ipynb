```python
import re
from docs.imaging_anatomy_list import IMAGING_ANATOMY_LIST
```

```python
basic_stop_words = [
    "a", "about", "an", "are", "as", "at", "be", "by", "com", "de", "en", "for", "from", "how", "i", "in", "is", "it", "la", "of", 
    "on", "or", "that", "the", "this", "to", "was", "what", "when", "where", "who", "will", "with", "und", "the", "www", 'non', 'only',
]
```

```python
def clean_string_edges(s):
    """
    Removes all special characters including spaces from the beginning and end of the string.
    Only keeps trimming at the edges; the content inside the string is untouched.
    """
    return re.sub(r'^[^a-zA-Z0-9]+|[^a-zA-Z0-9]+$', '', s)
```

```python
vocab_list = []

for anatomy_item in IMAGING_ANATOMY_LIST:
    # anatomy_words = anatomy_item.split()
    anatomy_words = re.split(' |/', anatomy_item)
    
    for word in anatomy_words:
        word = clean_string_edges(word)

        # ignore if remaining word is only a number
        # or a stop word
        if word.isdigit() or word in basic_stop_words:
            continue

        if len(word) > 3:
            word = word.lower()
            
        if len(word) > 1 and word not in vocab_list:
            vocab_list.append(word)
```
